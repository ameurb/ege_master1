### Why StandardScaler? / لماذا StandardScaler؟

**EN:** Economic variables have very different scales:
- `Inflation` might range from 0 to 25%
- `Government_Spending_GDP` might range from 8 to 42%
- `Oil_Revenue_Percent` might range from 0 to 58%

`StandardScaler` transforms each feature so that its **mean = 0** and **std = 1**. This prevents large-scale variables from dominating the model.

**Important rule:** Fit the scaler on **training data only**, then `transform` the test data — never fit on test data.

**AR:** المتغيرات الاقتصادية لها مقاييس مختلفة جداً. StandardScaler يحوّل كل متغير ليكون متوسطه 0 وانحرافه المعياري 1، مما يجعل المقارنة عادلة بين المتغيرات.

**قاعدة مهمة:** طبّق fit_transform على بيانات التدريب فقط، ثم transform فقط على بيانات الاختبار.
